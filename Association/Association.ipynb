```python
pivotted_UK_data.fillna(0,inplace = True)
one_hot_encoded_pivotted_UK_data = pivotted_UK_data.applymap(func = lambda x: 0 if x == 0 else 1)
```

```python
# 4. Data Mining
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
apriori(df = one_hot_encoded_pivotted_UK_data,min_support=0.02)
```

```python
frequent_item_sets=apriori(df = one_hot_encoded_pivotted_UK_data,min_support=0.02,use_colnames=True)
frequent_item_sets
```

```python
best_associates = association_rules(df=frequent_item_sets,metric='confidence',min_threshold=0.02)
best_associates
```
